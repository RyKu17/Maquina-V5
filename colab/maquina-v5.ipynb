```python
# ============================================
# CELDA 1: Montar Google Drive (opcional)
# ============================================
from google.colab import drive

MOUNT_DRIVE = False  # Cambiar a True si quieres backups

if MOUNT_DRIVE:
    drive.mount('/content/drive')
    print('Google Drive montado en /content/drive')
else:
    print('Google Drive NO montado (MOUNT_DRIVE = False)')
    print('Si quieres backups, cambia MOUNT_DRIVE = True y vuelve a ejecutar')
```

```python
# ============================================
# CELDA 2: Verificar region (opcional)
# ============================================
import requests, json

print('Verificando region del servidor...')
try:
    r = requests.get('https://ipinfo.io/json', timeout=10)
    data = r.json()
    print(f'Region: {data.get("region", "desconocida")}')
    print(f'Ciudad: {data.get("city", "desconocida")}')
    print(f'ISP: {data.get("org", "desconocida")}')
    print()
    print('Recomendacion: si la region no es optima,')
    print('ve a Runtime > Factory reset runtime y reconecta')
    print('para obtener una maquina virtual en otra region.')
except Exception as e:
    print(f'No se pudo verificar region: {e}')
```

```python
# ============================================
# CELDA 3: DESCARGA E INSTALACION
# ============================================
import os, subprocess, sys

REPO_URL = 'https://github.com/RyKu17/Maquina-V5'
SCRIPTS_DIR = '/content/Maquina-V5/scripts'

print('Descargando Maquina-V5...')
os.system(f'rm -rf /content/Maquina-V5 && git clone --depth=1 {REPO_URL} /content/Maquina-V5 2>/dev/null || echo "usando ZIP fallback"')

# Fallback: si no hay git, descargar ZIP
if not os.path.exists(SCRIPTS_DIR):
    os.system('wget -q -O /content/Maquina-V5.zip https://github.com/RyKu17/Maquina-V5/archive/refs/heads/main.zip')
    os.system('unzip -q /content/Maquina-V5.zip -d /content/')
    os.system('mv /content/Maquina-V5-main/* /content/Maquina-V5/ 2>/dev/null; rm -rf /content/Maquina-V5-main /content/Maquina-V5.zip')

if not os.path.exists(SCRIPTS_DIR):
    print('ERROR: No se pudieron descargar los scripts')
    sys.exit(1)

print('Scripts descargados correctamente')
print()

# Configuracion (editar si es necesario)
os.environ['SUNSHINE_USER'] = 'admin'
os.environ['SUNSHINE_PASS'] = 'admin'
os.environ['DISPLAY_RES'] = '1920x1080'
os.environ['BACKUP_ENABLED'] = 'true' if MOUNT_DRIVE else 'false'

print('Ejecutando instalador...')
os.system(f'bash {SCRIPTS_DIR}/install.sh')
print()
print('Instalacion completada. Revisa la IP de Tailscale arriba.')
```
